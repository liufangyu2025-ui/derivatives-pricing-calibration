```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))
```

```python
import numpy as np
import matplotlib.pyplot as plt

from src.models.black_scholes import (
    call_price,
    put_price,
    call_delta,
    put_delta,
    gamma,
    vega,
)
```

```python
K = 100
T = 1
r = 0.05
sigma = 0.20

spot_prices = np.linspace(50, 150, 200)
```

```python
call_prices = [
    call_price(S, K, T, r, sigma)
    for S in spot_prices
]
```

```python
plt.figure(figsize=(8, 5))
plt.plot(spot_prices, call_prices)

plt.xlabel("Spot Price")
plt.ylabel("European Call Price")
plt.title("Black-Scholes Call Price vs Spot Price")

plt.grid(True)
plt.show()
```

```python
call_deltas = [
    call_delta(S, K, T, r, sigma)
    for S in spot_prices
]

plt.figure(figsize=(8, 5))
plt.plot(spot_prices, call_deltas)

plt.xlabel("Spot Price")
plt.ylabel("Call Delta")
plt.title("Call Delta vs Spot Price")

plt.grid(True)
plt.show()
```

```python
gammas = [
    gamma(S, K, T, r, sigma)
    for S in spot_prices
]

plt.figure(figsize=(8, 5))
plt.plot(spot_prices, gammas)

plt.xlabel("Spot Price")
plt.ylabel("Gamma")
plt.title("Gamma vs Spot Price")

plt.grid(True)
plt.show()
```
